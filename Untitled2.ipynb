```python
class PDBReader:

    def __init__(self, filepath):
        self.pdbName = os.path.basename(filepath)
        self.coordinates = {}
        self.residue_order = {}
        self.metadata = {"protein_name": "not found",
            "organism": "not found",
            "other_info": [] }
        self._read_pdb(filepath)

    def _read_pdb(self, filepath):
        with open(filepath, "r") as f:
            for line in f:
                rec = line[:6].strip()

                if rec == "TITLE":
                    name = line[10:].strip()
                    if self.metadata["protein_name"] == "not found":
                        self.metadata["protein_name"] = name
                    else:
                        self.metadata["protein_name"] += " " + name
                elif rec == "SOURCE":
                    content = line[10:].strip()
                    if "ORGANISM_SCIENTIFIC" in content or "ORGANISM:" in content:
                        parts = content.split(":")
                        if len(parts) > 1:
                            org = parts[-1].strip().rstrip(";")
                            if org:
                                self.metadata["organism"] = org
                    self.metadata["other_info"].append(content)

                elif rec == "HEADER":
                    self.metadata["other_info"].append(line[10:].strip())

                elif rec == "ATOM":
                    try:
                        atom_name = line[12:16].strip()
                        chain_id  = line[21].strip()
                        res_num   = int(line[22:26].strip())
                        x = float(line[30:38].strip())
                        y = float(line[38:46].strip())
                        z = float(line[46:54].strip())

                        if chain_id not in self.coordinates:
                            self.coordinates[chain_id] = {}
                            self.residue_order[chain_id] = []

                        if res_num not in self.coordinates[chain_id]:
                            self.coordinates[chain_id][res_num] = {}
                            self.residue_order[chain_id].append(res_num)

                        self.coordinates[chain_id][res_num][atom_name] = (x, y, z)

                    except:
                        pass
    def ListRes(self, chain_id):
        if chain_id not in self.coordinates:
            print(f"chain '{chain_id}' not found. available chains: {list(self.coordinates.keys())}")
            return []
        return list(self.residue_order[chain_id])

    def atoms(self, chain_id, resID):
        if chain_id not in self.coordinates:
            print(f"chain '{chain_id}' not found")
            return []
        if resID not in self.coordinates[chain_id]:
            print(f"residue {resID} not found in chain '{chain_id}'")
            return []
        return list(self.coordinates[chain_id][resID].keys())

    def getCord(self, chain_id, resID, atom_name):
        if chain_id not in self.coordinates:
            print(f"chain '{chain_id}' not found. available: {list(self.coordinates.keys())}")
            return None
        if resID not in self.coordinates[chain_id]:
            print(f"residue {resID} not in chain '{chain_id}'")
            return None
        if atom_name not in self.coordinates[chain_id][resID]:
            print(f"atom '{atom_name}' not found. available: {list(self.coordinates[chain_id][resID].keys())}")
            return None
        return self.coordinates[chain_id][resID][atom_name]

    def findHBond(self, chain_id, i, j):
        if chain_id not in self.coordinates:
            print(f"chain '{chain_id}' not found")
            return False
        if i == j:
            print("i and j must be different residues")
            return False

        res = self.coordinates[chain_id]

        if i not in res or j not in res:
            return False
        if "N" not in res[i] or "O" not in res[j] or "C" not in res[j]:
            return False

        N = res[i]["N"]
        O = res[j]["O"]
        C = res[j]["C"]

        if self._dist(N, O) >= 3.8:
            return False

        angle = self._angle(N, O, C)
        if not (120.0 <= angle <= 180.0):
            return False
        return True

    def _dist(self, p1, p2):
        return math.sqrt(sum((a - b)**2 for a, b in zip(p1, p2)))

    def _angle(self, A, B, C):
        BA = [a - b for a, b in zip(A, B)]
        BC = [c - b for c, b in zip(C, B)]
        dot = sum(x * y for x, y in zip(BA, BC))
        mag_BA = math.sqrt(sum(x**2 for x in BA))
        mag_BC = math.sqrt(sum(x**2 for x in BC))
        if mag_BA == 0 or mag_BC == 0:
            return 0.0
        cos_val = max(-1.0, min(1.0, dot / (mag_BA * mag_BC)))
        return math.degrees(math.acos(cos_val))

    def count_all_hbonds(self, chain_id):
        residues = self.ListRes(chain_id)
        total = 0
        for i in residues:
            for j in residues:
                if i != j and self.findHBond(chain_id, i, j):
                    total += 1
        return total

print("class defined")
```

```python
uniprot_ids = ["P0ABT2", "Q8XF78", "Q2NUI3", "B6I7W9", "B4TQX5", "Q57RC9"]

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "application/json, text/plain, */*",
}

def get_pdb_url(uid):
    api_url = f"https://alphafold.ebi.ac.uk/api/prediction/{uid}"
    resp = requests.get(api_url, headers=headers, timeout=30)
    if resp.status_code != 200:
        print(f"  API error {resp.status_code} for {uid}")
        return None
    data = resp.json()
    if not data:
        return None
    return data[0].get("pdbUrl")

def download_pdb(uid):
    filename = f"AF_{uid}.pdb"

    if os.path.exists(filename):
        print(f"{uid} already downloaded")
        return filename

    print(f"{uid}: getting download link...", end=" ")
    url = get_pdb_url(uid)
    if not url:
        print("failed to get url")
        return None
    print("got it")

    print(f"{uid}: downloading...", end=" ")
    resp = requests.get(url, headers=headers, timeout=60)
    if resp.status_code != 200:
        print(f"download failed {resp.status_code}")
        return None

    text = resp.content.decode("utf-8", errors="ignore")
    if "ATOM" not in text:
        print("no ATOM records, skipping")
        return None

    with open(filename, "w") as f:
        f.write(text)

    print(f"saved as {filename}")
    return filename

downloaded = {}
for uid in uniprot_ids:
    path = download_pdb(uid)
    if path:
        downloaded[uid] = path

print(f"\ndownloaded {len(downloaded)}/{len(uniprot_ids)} files")
```

```python

uniprot_ids = ["P0ABT2", "Q8XF78", "Q2NUI3", "B6I7W9", "B4TQX5", "Q57RC9"]

ALPHAFOLD_API = "https://alphafold.ebi.ac.uk/api/prediction/{uid}"

HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/120.0.0.0 Safari/537.36"
    ),
    "Accept": "application/json, text/plain, */*",
}


def get_pdb_url_from_api(uid):
    """
    Call the AlphaFold API for a UniProt ID and return the pdbUrl.
    Returns the URL string on success, or None on failure.
    """
    api_url = ALPHAFOLD_API.format(uid=uid)
    try:
        resp = requests.get(api_url, headers=HEADERS, timeout=30)
        if resp.status_code != 200:
            print(f"  API returned HTTP {resp.status_code}")
            return None
        data = resp.json()
        # API returns a list; take the first entry
        if not data or not isinstance(data, list):
            print(f"  API returned unexpected format: {data}")
            return None
        pdb_url = data[0].get("pdbUrl")
        if not pdb_url:
            print(f"  No 'pdbUrl' found in API response")
            return None
        return pdb_url
    except Exception as exc:
        print(f"  API call error: {exc}")
        return None


def download_one(uid, save_dir="."):
    """
    Download the AlphaFold PDB file for a single UniProt ID.
    Uses the API to discover the correct pdbUrl first.
    Returns the local filepath on success, or None on failure.
    """
    filename = os.path.join(save_dir, f"AF_{uid}.pdb")

    if os.path.exists(filename):
        print(f"[{uid}]  ✓ already on disk → {filename}")
        return filename

    print(f"[{uid}]  Step 1 — querying AlphaFold API ...", end=" ", flush=True)
    pdb_url = get_pdb_url_from_api(uid)
    if not pdb_url:
        print(f"[{uid}]  ✗ Could not get pdbUrl from API")
        return None
    print(f"OK  →  {pdb_url.split('/')[-1]}")

    print(f"[{uid}]  Step 2 — downloading PDB ...", end=" ", flush=True)
    try:
        resp = requests.get(pdb_url, headers=HEADERS, timeout=60)
        if resp.status_code != 200:
            print(f"HTTP {resp.status_code}")
            return None

        text = resp.content.decode("utf-8", errors="ignore")

        if "ATOM" not in text:
            print("downloaded file has no ATOM records — skipping")
            return None

        with open(filename, "w") as fh:
            fh.write(text)

        size_kb = os.path.getsize(filename) / 1024
        print(f"✓ saved ({size_kb:.0f} KB) → {filename}")
        return filename

    except Exception as exc:
        print(f"error: {exc}")
        return None


print("=" * 60)
print("Downloading AlphaFold PDB files via API — one at a time")
print("=" * 60)

downloaded_files = {}

for uid in uniprot_ids:
    path = download_one(uid)
    if path:
        downloaded_files[uid] = path
    print()

print(f"Result: {len(downloaded_files)} / {len(uniprot_ids)} downloaded successfully")
print("Files :", list(downloaded_files.values()))

```

```python
hbond_counts = {}

print("counting hydrogen bonds...\n")
print(f"{'UniProt ID':<12}  {'# Residues':>10}  {'H-Bonds':>8}")
print("-" * 35)

for uid, filepath in downloaded.items():
    pdb = PDBReader(filepath)
    residues = pdb.ListRes("A")
    hbonds = pdb.count_all_hbonds("A")
    hbond_counts[uid] = hbonds
    print(f"{uid:<12}  {len(residues):>10}  {hbonds:>8}")

avg_hbonds = sum(hbond_counts.values()) / len(hbond_counts)
print("-" * 35)
print(f"{'Average':<12}  {'':>10}  {avg_hbonds:>8.1f}")
```

```python
labels = list(hbond_counts.keys())
counts = [hbond_counts[uid] for uid in labels]
avg = sum(counts) / len(counts)

fig, ax = plt.subplots(figsize=(9, 5))

bars = ax.bar(labels, counts, color="#5b8dd9", edgecolor="white", width=0.5)

for bar, count in zip(bars, counts):
    if count > avg:
        bar.set_color("#2c5f9e")
ax.axhline(avg, color="red", linestyle="--", linewidth=1.5, label=f"Average = {avg:.1f}")

for bar, count in zip(bars, counts):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + max(counts) * 0.01,
        str(count),
        ha="center", va="bottom", fontsize=9, fontweight="bold"
    )

ax.set_xlabel("UniProt ID")
ax.set_ylabel("Number of Main-Chain H-Bonds")
ax.set_title("Main-Chain H-Bonds in AlphaFold Structures\n(N-O dist < 3.8Å, N-O-C angle 120-180°)")
ax.legend()
ax.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.savefig("hbond_plot.png", dpi=150)
plt.show()
print("plot saved")
```
